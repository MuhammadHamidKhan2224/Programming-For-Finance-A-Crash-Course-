# 1. fv (Future Value)

### Calculates the future value of an investment based on a constant interest rate and regular payments. This helps you figure out how much your savings will be worth after a specific number of years.

# 2. pv (Present Value)

## Calculates the present value of a future sum of money or stream of cash flows given a specific rate of return. It tells you what future money is worth in today's dollars.

# 3. npv (Net Present Value)

## Calculates the sum of present values for a series of irregular, varying cash flows (like an initial investment followed by different yearly returns). It is used to determine if a business project is profitable.

# 4. pmt (Payment)

## Computes the total fixed payment (principal + interest) needed to pay off a loan over a set number of periods at a constant interest rate.

# 5. ppmt (Principal Payment)

## Calculates just the principal portion of a specific payment period. This shows how much of your payment is actually reducing the loan balance during a given month.

# 6. ipmt (Interest Payment)

## Calculates just the interest portion of a specific payment period. This shows how much of your payment is going directly to the lender as profit rather than paying down your debt.

# 7. nper (Number of Periods)

## Determines how many payment periods it will take to pay off a loan or reach an investment goal, given a fixed payment amount and interest rate.

# 8. rate (Interest Rate)

## Computes the interest rate per period of an annuity or loan. This is useful if you know the loan amount, your monthly payment, and the duration, but the lender didn't explicitly state the interest rate.

# 9. irr (Internal Rate of Return)

## Calculates the rate of return for a series of periodic cash flows. It represents the exact interest rate that makes the Net Present Value (NPV) of those cash flows equal to zero.

## 10. mirr (Modified Internal Rate of Return)

## Similar to IRR, but more realistic: it accounts for the fact that positive cash flows are reinvested at a specific rate, and negative cash flows are financed at a different borrowing rate.

# Part A — Bond Valuation

## Question 1: Annual-Coupon Bond Valuation

```python
# Given data
# Face value: PKR 1,000
# Annual coupon rate: 9%
# Required return: 11%
# Remaining maturity: 6 years
# Coupons paid annually

# Calculate the annual coupon, present value of coupons, present value of face value, total bond price and valuation status.
```

```python
import numpy_financial as npf
```

```python
# Given data
face_value = 1000
coupon_rate = 0.09
required_return = 0.11
years = 6

# Annual coupon payment
annual_coupon = face_value * coupon_rate
print(annual_coupon)

# Present value of coupon payments
pv_coupons = -npf.pv(
    rate=required_return,
    nper=years,
    pmt=annual_coupon,
    fv=0
)

# Present value of face value
pv_face_value = -npf.pv(
    rate=required_return,
    nper=years,
    pmt=0,
    fv=face_value
)

# Total bond price
bond_price = pv_coupons + pv_face_value

# Determine valuation status
if bond_price > face_value:
    bond_status = "Premium Bond"
elif bond_price < face_value:
    bond_status = "Discount Bond"
else:
    bond_status = "Par Bond"

# Display results
print("--- Annual-Coupon Bond Valuation ---")
print("Annual Coupon: PKR", round(annual_coupon, 2))
print("PV of Coupons: PKR", round(pv_coupons, 2))
print("PV of Face Value: PKR", round(pv_face_value, 2))
print("Current Bond Price: PKR", round(bond_price, 2))
print("Bond Status:", bond_status)
```

## Question 2: Semiannual-Coupon Bond Valuation

```python
# Given data
# Face value: PKR 5,000
# Annual coupon rate: 10% (0.10)
# Required annual return: 8% (0.08)
# Maturity: 4 years
# Coupons paid semiannually
# Calculate the semiannual coupon, periodic return, number of periods, bond price and premium or discount amount.
```

```python
# Given data
face_value = 5000
annual_coupon_rate = 0.10
annual_required_return = 0.08
years = 4
payments_per_year = 2

# Adjust values for semiannual payments
semiannual_coupon = (face_value * annual_coupon_rate) / payments_per_year
periodic_return = (annual_required_return / payments_per_year)

number_of_periods = years * payments_per_year

# Calculate bond price
bond_price = -npf.pv(
    rate=periodic_return,
    nper=number_of_periods,
    pmt=semiannual_coupon,
    fv=face_value
)

# Calculate premium or discount
difference = bond_price - face_value

if difference > 0:
    bond_status = "Premium Bond"
    valuation_amount = difference
elif difference < 0:
    bond_status = "Discount Bond"
    valuation_amount = abs(difference)
else:
    bond_status = "Par Bond"
    valuation_amount = 0

# Display results
print("--- Semiannual Bond Valuation ---")
print("Semiannual Coupon: PKR", round(semiannual_coupon, 2))
print("Periodic Return:", round(periodic_return * 100, 2), "%")
print("Total Payment Periods:", number_of_periods)
print("Current Bond Price: PKR", round(bond_price, 2))
print("Bond Status:", bond_status)
print("Premium/Discount Amount: PKR", round(valuation_amount, 2))
```

# Question 3: Bond Price Sensitivity

```python
# Given data
# Face value: PKR 1,000
# Annual coupon rate: 10%
# Maturity: 5 years
# # Coupons paid annually
# Required returns: 8%, 10% and 12%
# Calculate the bond price under each required return and explain the relationship between required return and bond price.
```

```python
# Given data
face_value = 1000
coupon_rate = 0.10
years = 5

# Annual coupon payment
annual_coupon = face_value * coupon_rate

# Required returns
return_1 = 0.08
return_2 = 0.10
return_3 = 0.12

# Bond price at 8% required return
price_at_8_percent = -npf.pv(
    rate=return_1,
    nper=years,
    pmt=annual_coupon,
    fv=face_value
)

# Bond price at 10% required return
price_at_10_percent = -npf.pv(
    rate=return_2,
    nper=years,
    pmt=annual_coupon,
    fv=face_value
)

# Bond price at 12% required return
price_at_12_percent = -npf.pv(
    rate=return_3,
    nper=years,
    pmt=annual_coupon,
    fv=face_value
)

# Display results
print("--- Bond Price Sensitivity Analysis ---")
print("Annual Coupon: PKR", round(annual_coupon, 2))
print("Price at 8% Return: PKR", round(price_at_8_percent, 2))
print("Price at 10% Return: PKR", round(price_at_10_percent, 2))
print("Price at 12% Return: PKR", round(price_at_12_percent, 2))

print("\nInterpretation:")
print("At 8% required return, the bond sells at a premium.")
print("At 10% required return, the bond sells at par.")
print("At 12% required return, the bond sells at a discount.")
print("Bond price decreases when the required return increases.")
```

# Question 4: Calculating Yield to Maturity

```python
# Given data
# Current market price: PKR 950
# Face value: PKR 1,000
# Annual coupon payment: PKR 80
# Remaining maturity: 5 years
# Coupons paid annually
# Calculate the annual yield to maturity and compare it with the coupon rate.
```

```python
# Given data
market_price = 950
face_value = 1000
annual_coupon = 80
years = 5

# Coupon rate
coupon_rate = annual_coupon / face_value

# Calculate yield to maturity
ytm = npf.rate(
    nper=years,
    pmt=annual_coupon,
    pv=-market_price,
    fv=face_value
)

# Determine bond status
if market_price > face_value:
    bond_status = "Premium Bond"
elif market_price < face_value:
    bond_status = "Discount Bond"
else:
    bond_status = "Par Bond"

# Display results
print("--- Yield to Maturity Calculation ---")
print("Market Price: PKR", market_price)
print("Face Value: PKR", face_value)
print("Coupon Rate:", round(coupon_rate * 100, 2), "%")
print("Yield to Maturity:", round(ytm * 100, 2), "%")
print("Bond Status:", bond_status)

if ytm > coupon_rate:
    print("The YTM is higher than the coupon rate.")
elif ytm < coupon_rate:
    print("The YTM is lower than the coupon rate.")
else:
    print("The YTM is equal to the coupon rate.")
```

# Question 5: Revaluation of an Existing Bond

```python
# Given data
# Original purchase price: PKR 2,000
# Face value: PKR 2,000
# Annual coupon rate: 7%
# Remaining maturity: 6 years
# Current required return: 9%
# Coupons paid annually
# Calculate the current bond value, capital gain or loss, and percentage gain or loss.
```

```python
# Given data
original_purchase_price = 2000
face_value = 2000
coupon_rate = 0.07
required_return = 0.09
remaining_years = 6

# Annual coupon payment
annual_coupon = face_value * coupon_rate

# Current bond value
current_bond_value = -npf.pv(
    rate=required_return,
    nper=remaining_years,
    pmt=annual_coupon,
    fv=face_value
)

# Capital gain or loss
capital_gain_loss = (
    current_bond_value - original_purchase_price
)

# Percentage gain or loss
percentage_change = (
    capital_gain_loss / original_purchase_price
) * 100

# Determine result
if capital_gain_loss > 0:
    investment_result = "Capital Gain"
elif capital_gain_loss < 0:
    investment_result = "Capital Loss"
else:
    investment_result = "No Gain or Loss"

# Display results
print("--- Existing Bond Revaluation ---")
print("Annual Coupon: PKR", round(annual_coupon, 2))
print("Original Price: PKR", original_purchase_price)
print("Current Bond Value: PKR", round(current_bond_value, 2))
print("Gain/Loss Amount: PKR", round(capital_gain_loss, 2))
print("Percentage Change:", round(percentage_change, 2), "%")
print("Investment Result:", investment_result)
```

## Question 6: One-Period Stock Valuation

```python
# Given data
# Expected dividend next year: PKR 7.50
# Expected selling price after one year: PKR 150
# Required return: 14%
# Current market price: PKR 132
# Calculate the intrinsic value and determine whether the stock is undervalued or overvalued.
```

```python
# Given data
expected_dividend = 7.50
expected_selling_price = 150
required_return = 0.14
market_price = 132

# Total expected cash flow after one year
future_cash_flow = (expected_dividend + expected_selling_price)

# Present value of future cash flow
intrinsic_value = -npf.pv(
    rate=required_return,
    nper=1,
    pmt=0,
    fv=future_cash_flow
)

# Difference between intrinsic value and market price
price_difference = intrinsic_value - market_price

# Determine valuation status
if intrinsic_value > market_price:
    stock_status = "Undervalued"
elif intrinsic_value < market_price:
    stock_status = "Overvalued"
else:
    stock_status = "Fairly Valued"

# Display results
print("--- One-Period Stock Valuation ---")
print("Expected Dividend: PKR", expected_dividend)
print("Expected Selling Price: PKR", expected_selling_price)
print("Total Future Cash Flow: PKR", future_cash_flow)
print("Intrinsic Value: PKR", round(intrinsic_value, 2))
print("Current Market Price: PKR", market_price)
print("Price Difference: PKR", round(price_difference, 2))
print("Stock Status:", stock_status)
```

## Question 7: Three-Year Stock Valuation

```python
# Given data
# Dividend in Year 1: PKR 4
# Dividend in Year 2: PKR 5
# Dividend in Year 3: PKR 6
# Expected selling price after Year 3: PKR 120
# Required return: 11%
# Calculate the present value of every cash flow and the total intrinsic stock value.
```

```python
# Given data
dividend_1 = 4
dividend_2 = 5
dividend_3 = 6
selling_price = 120
required_return = 0.11

# Total cash flow in Year 3
year_3_cash_flow = dividend_3 + selling_price

# Present value of individual cash flows
pv_dividend_1 = dividend_1 / (1 + required_return) ** 1
pv_dividend_2 = dividend_2 / (1 + required_return) ** 2
pv_year_3_cash_flow = (
    year_3_cash_flow / (1 + required_return) ** 3
)

# Cash-flow list for NPV
cash_flows = [
    0,
    dividend_1,
    dividend_2,
    year_3_cash_flow
]

# Intrinsic stock value
intrinsic_value = npf.npv(
    required_return,
    cash_flows
)

# Display results
print("--- Three-Year Stock Valuation ---")
print("PV of Year 1 Dividend: PKR", round(pv_dividend_1, 2))
print("PV of Year 2 Dividend: PKR", round(pv_dividend_2, 2))
print("PV of Year 3 Cash Flow: PKR", round(pv_year_3_cash_flow, 2))
print("Intrinsic Stock Value: PKR", round(intrinsic_value, 2))
```

# Question 8: Constant-Growth Stock Valuation

```python
# Given data
# Dividend just paid: PKR 6
# Expected annual dividend growth: 4%
# Required return: 11%
# Current market price: PKR 95
# Calculate the expected dividend next year, intrinsic value and stock valuation status.
```

```python
# Given data
current_dividend = 6
growth_rate = 0.04
required_return = 0.11
market_price = 95

# Calculate next year's dividend
next_dividend = current_dividend * (1 + growth_rate)

# Gordon Growth Model
intrinsic_value = next_dividend / (
    required_return - growth_rate
)

# Difference between intrinsic value and market price
price_difference = intrinsic_value - market_price

# Determine valuation status
if intrinsic_value > market_price:
    stock_status = "Undervalued"
elif intrinsic_value < market_price:
    stock_status = "Overvalued"
else:
    stock_status = "Fairly Valued"

# Display results
print("--- Constant-Growth Stock Valuation ---")
print("Current Dividend: PKR", current_dividend)
print("Next Year's Dividend: PKR", round(next_dividend, 2))
print("Intrinsic Stock Value: PKR", round(intrinsic_value, 2))
print("Current Market Price: PKR", market_price)
print("Price Difference: PKR", round(price_difference, 2))
print("Stock Status:", stock_status)
```

# Question 9: Two-Stage Dividend Growth Model

```python
# Given data
# Dividend just paid: PKR 5
# Growth rate for the first three years: 12%
# Stable growth rate after Year 3: 5%
# Required return: 13%
# Calculate the dividends for Years 1–4, terminal value and intrinsic stock value.
```

```python
# Given data
current_dividend = 5
high_growth_rate = 0.12
stable_growth_rate = 0.05
required_return = 0.13

# Dividends during high-growth period
dividend_1 = current_dividend * (1 + high_growth_rate)
dividend_2 = dividend_1 * (1 + high_growth_rate)
dividend_3 = dividend_2 * (1 + high_growth_rate)

# First dividend in stable-growth period
dividend_4 = dividend_3 * (1 + stable_growth_rate)

# Terminal value at the end of Year 3
terminal_value = dividend_4 / (
    required_return - stable_growth_rate
)

# Present value of dividends
pv_dividend_1 = dividend_1 / (1 + required_return) ** 1
pv_dividend_2 = dividend_2 / (1 + required_return) ** 2
pv_dividend_3 = dividend_3 / (1 + required_return) ** 3

# Present value of terminal value
pv_terminal_value = (
    terminal_value / (1 + required_return) ** 3
)

# Cash flows for NPV
cash_flows = [
    0,
    dividend_1,
    dividend_2,
    dividend_3 + terminal_value
]

# Intrinsic stock value
intrinsic_value = npf.npv(
    required_return,
    cash_flows
)

# Display results
print("--- Two-Stage Dividend Growth Model ---")
print("Dividend in Year 1: PKR", round(dividend_1, 2))
print("Dividend in Year 2: PKR", round(dividend_2, 2))
print("Dividend in Year 3: PKR", round(dividend_3, 2))
print("Dividend in Year 4: PKR", round(dividend_4, 2))
print("Terminal Value in Year 3: PKR", round(terminal_value, 2))
print("PV of Year 1 Dividend: PKR", round(pv_dividend_1, 2))
print("PV of Year 2 Dividend: PKR", round(pv_dividend_2, 2))
print("PV of Year 3 Dividend: PKR", round(pv_dividend_3, 2))
print("PV of Terminal Value: PKR", round(pv_terminal_value, 2))
print("Intrinsic Stock Value: PKR", round(intrinsic_value, 2))
```

## Question 10: Expected Stock Return Using IRR

```python
# Given data
# Current purchase price: PKR 90
# Dividend in Year 1: PKR 4
# Dividend in Year 2: PKR 4.50
# Dividend in Year 3: PKR 5
# Expected selling price after Year 3: PKR 110
# Minimum required return: 10%
# Calculate the expected annual return and determine whether the investment is acceptable.
```

```python
# Given data
purchase_price = 90
dividend_1 = 4
dividend_2 = 4.50
dividend_3 = 5
selling_price = 110
minimum_required_return = 0.10

# Total cash flow in Year 3
year_3_cash_flow = dividend_3 + selling_price

# Investment cash flows
cash_flows = [
    -purchase_price,
    dividend_1,
    dividend_2,
    year_3_cash_flow
]

# Calculate expected annual return
expected_return = npf.irr(cash_flows)

# Investment decision
if expected_return >= minimum_required_return:
    investment_decision = "Accept the Investment"
else:
    investment_decision = "Reject the Investment"

# Display results
print("--- Expected Stock Return ---")
print("Purchase Price: PKR", purchase_price)
print("Year 1 Dividend: PKR", dividend_1)
print("Year 2 Dividend: PKR", dividend_2)
print("Year 3 Total Cash Flow: PKR", year_3_cash_flow)
print("Expected Annual Return:", round(expected_return * 100, 2), "%")
print(
    "Minimum Required Return:",
    round(minimum_required_return * 100, 2),
    "%"
)
print("Investment Decision:", investment_decision)
```

# 2. Calculate expected cash flow

```python
probab = [0.08,0.25,0.45,0.25,0.10]
cash_flow_project_A = [5000, 6000, 7000, 8000, 9000]
cash_flow_project_B = [6000, 7000, 8000, 9000, 1000]

if not (len(probab) == len(cash_flow_project_A) == len(cash_flow_project_B)):
  raise ValueError("Match the values size is each list above")
if any(p < 0 for p in probs) or not np.isclose(sum(probs), 1.0):
    raise ValueError("Probabilities must be non-negative and add to 1.")
ev_a = 0.0
ev_b = 0.0

for i ,p  in enumerate(probab):
  ev_a += cash_flow_project_A[i] * p
  ev_b += cash_flow_project_B[i] * p

print(round(ev_a, 2), round(ev_b, 2))
```

```python
var_pro_a = 0.0
var_pro_b = 0.0

for i , p in enumerate (probab):
  var_pro_a += (cash_flow_project_A[i] - ev_a) **2 * p
  var_pro_b += (cash_flow_project_B[i] - ev_a) **2 * p
```

# 4. Check the calculation with NumPy

```python
probs_np = np.array(probs)
cf_b_np = np.array(cf_b)

if not np.isclose(probs_np.sum(), 1.0):
    raise ValueError("Probabilities must add to 1.")

ev_b_np = probs_np @ cf_b_np
sd_b_np = np.sqrt(probs_np @ (cf_b_np - ev_b_np) ** 2)

print(round(ev_b_np, 2), round(sd_b_np, 2), round(sd_b_np / ev_b_np, 4))
```

# 5. Represent the probability tree as rows

```python
# Each row is: CF1, CF2, P(1), P(2 given 1)
tree = [
    [ 500,  800, 0.25, 0.40],
    [ 500,  500, 0.25, 0.40],
    [ 500,  200, 0.25, 0.20],
    [ 200,  500, 0.50, 0.20],
    [ 200,  200, 0.50, 0.60],
    [ 200, -100, 0.50, 0.20],
    [-100,  200, 0.25, 0.20],
    [-100, -100, 0.25, 0.40],
    [-100, -400, 0.25, 0.40],
]

joint_total = 0.0

for cf1, cf2, p1, p2 in tree:
    joint_total += p1 * p2

if not np.isclose(joint_total, 1.0):
    raise ValueError("Joint path probabilities must add to 1.")

print(round(joint_total, 2))
```

# 6. Calculate expected NPV for the tree

```python
risk_free_rate = 0.08
initial_outflow = -240
expected_npv = 0.0

for cf1, cf2, p1, p2 in tree:
    joint_probability = p1 * p2
    path_npv = npf.npv(risk_free_rate, [initial_outflow, cf1, cf2])
    expected_npv += joint_probability * path_npv

print(round(expected_npv, 2))
```

# 7. Calculate the standard deviation of NPV

```python
variance_npv = 0.0

for cf1, cf2, p1, p2 in tree:
    joint_probability = p1 * p2
    path_npv = npf.npv(risk_free_rate, [initial_outflow, cf1, cf2])
    variance_npv += (path_npv - expected_npv) ** 2 * joint_probability

sd_npv = variance_npv ** 0.5

print(round(variance_npv), round(sd_npv, 2))
```

# 8. Estimate the chance of a loss

```python
z = (0 - expected_npv) / sd_npv
probability_of_loss = NormalDist().cdf(z)

print(round(z, 2))
# -0.26

print(f"{probability_of_loss:.4f}")
```

# 10. Calculate risk for a two-project portfolio

```python
project_sd = [14000, 6000]
correlation = [
    [1.0, 0.4],
    [0.4, 1.0]
]

portfolio_variance = 0.0

for j in range(2):
    for k in range(2):
        portfolio_variance += (
            correlation[j][k] * project_sd[j] * project_sd[k]
        )

portfolio_sd = portfolio_variance ** 0.5

print(round(portfolio_sd))
# 17297

print(round(14000 / 12000, 2), round(portfolio_sd / 20000, 2))
```

# 11. Find dominated project choices

```python
plaza = {
    "A": (100, 200), "B": (20, 80),  "C": (75, 100),
    "D": (60, 150),  "E": (50, 20),  "F": (40, 60),
    "G": (120, 170), "H": (90, 70),  "I": (50, 100),
    "J": (75, 30)
}
# Each value is (expected NPV, standard deviation), in $000s.

survivors = []

for name, (mean, sd) in plaza.items():
    dominated = False

    for other, (other_mean, other_sd) in plaza.items():
        if (
            other != name
            and other_mean >= mean
            and other_sd <= sd
            and (other_mean > mean or other_sd < sd)
        ):
            dominated = True
            break

    if not dominated:
        survivors.append(name)

print(survivors)
```
